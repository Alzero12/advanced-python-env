```python
import matplotlib.pyplot as plt
import pandas as pd

names = ["Ali", "Aruzhan", "Daniyar", "Amina", "Timur"]
scores = [85, 90, 78, 92, 88]

plt.plot(names, scores)
plt.xlabel("Students")
plt.ylabel("Scores")
plt.title("Student Scores Line Plot")
plt.show()

plt.bar(names, scores)
plt.xlabel("Students")
plt.ylabel("Scores")
plt.title("Student Scores Bar Chart")
plt.show()
```
